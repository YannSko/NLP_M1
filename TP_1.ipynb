## Gensim


```python
from gensim.test.utils import datapath
from gensim.models.word2vec import Text8Corpus

# Chemin vers le corpus
import gensim.downloader as api
dataset = api.load("text8")  # Text8 est téléchargé automatiquement ici
```

```python
from gensim.models import Word2Vec

# Entraîner le modèle Word2Vec (ça peut prendre 1-2 minutes)
model = Word2Vec(sentences=dataset, vector_size=100, window=5, min_count=5, workers=4)

# Sauvegarder le modèle (optionnel)
# model.save("word2vec_text8.model")
```

```python
print("Mots similaires à 'language' :")
print(model.wv.most_similar("language"))
```

```python
print("Similarité entre 'machine' et 'learning' :")
print(model.wv.similarity("machine", "learning"))
```

```python
print("king - man + woman =")
print(model.wv.most_similar(positive=["king", "woman"], negative=["man"]))
```

### TF IDF


```python
from sklearn.feature_extraction.text import TfidfVectorizer
import gensim.downloader as api
dataset = list(api.load("text8"))  


# Pour simplifier, on prend les 1000 premières phrases du corpus
corpus = [' '.join(sent) for sent in dataset[:1000]]

# Initialiser le vectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# Obtenir les mots du vocabulaire
feature_names = tfidf_vectorizer.get_feature_names_out()
```

```python
print("Word2Vec - Similaire à 'language':")
print(model.wv.most_similar("language"))
```

```python
import numpy as np

# Chercher l’index du mot 'language'
word_idx = tfidf_vectorizer.vocabulary_.get("language")
if word_idx is not None:
    # Trouver le document (phrase) avec le plus grand score pour 'language'
    best_doc_idx = np.argmax(tfidf_matrix[:, word_idx].toarray())
    row = tfidf_matrix[best_doc_idx].toarray().flatten()
    top_indices = row.argsort()[-5:][::-1]
    top_words = [feature_names[i] for i in top_indices]
    print("TF-IDF - Mots les plus importants dans la phrase avec 'language':")
    print(top_words)
else:
    print("'language' n'est pas dans le vocabulaire TF-IDF.")
```

```python
print("Word2Vec - Similarité machine/learning:")
print(model.wv.similarity("machine", "learning"))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Vérifier si les mots existent
i1 = tfidf_vectorizer.vocabulary_.get("machine")
i2 = tfidf_vectorizer.vocabulary_.get("learning")

if i1 is not None and i2 is not None:
    # Extraire les colonnes
    vec1 = tfidf_matrix[:, i1]
    vec2 = tfidf_matrix[:, i2]
    sim = cosine_similarity(vec1.T, vec2.T)
    print("TF-IDF - Similarité machine/learning :", sim[0][0])
else:
    print("Un des mots n'existe pas dans le vocabulaire TF-IDF.")
```

### Représentation graphique


```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Choisir 100 mots les plus fréquents du vocabulaire appris
words = list(model.wv.index_to_key[:100])  # Top 100 mots
vectors = np.array([model.wv[word] for word in words])  # Leurs vecteurs
```

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
vectors_2d = tsne.fit_transform(vectors)
```

```python
plt.figure(figsize=(12, 8))
plt.scatter(vectors_2d[:, 0], vectors_2d[:, 1])

# Ajouter les noms des mots
for i, word in enumerate(words):
    plt.annotate(word, (vectors_2d[i, 0], vectors_2d[i, 1]))

plt.title("Word2Vec – Visualisation 2D des mots avec t-SNE")
plt.grid(True)
plt.show()
```
